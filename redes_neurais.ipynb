```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics import confusion_matrix
from keras.models import Sequential
from keras.layers import Dense, Dropout
from google.colab import files
```

```python
spam = pd.read_csv("spam.csv")
spam.head()
```

```python
# fit = cria o modelo
# transform = converte texto para número

label_encoder = LabelEncoder()
y = label_encoder.fit_transform(spam['Category'])
print(y)
```

```python
mensagens = spam['Message'].values
X_train, X_test, y_train, y_test = train_test_split(mensagens, y, test_size=0.3)
print(X_train)
```

```python
# fit apenas no teste, transform em ambos

vetorizador = CountVectorizer()
vetorizador.fit(X_train)
X_train = vetorizador.transform(X_train)

X_test = vetorizador.transform(X_test)

print(X_train.toarray())
```

```python
import numpy as np
np.set_printoptions(threshold=np.inf)

X_train.toarray()[0]
```

```python
print(X_test.shape[1])
```

O código abaixo define a arquitetura de uma rede neural artificial sequencial para classificação. Aqui está o que cada linha faz:

1. `modelo = Sequential()`: Cria um modelo onde as camadas são empilhadas em ordem.

2. Primeira Camada (Dense):
- `units=10`: Tem 10 neurônios.
- `activation='relu'`: Usa a função ReLU para aprender padrões complexos.
- `input_dim=X_train.shape[1]`: Define a entrada com base no número de palavras (colunas) geradas pelo CountVectorizer.

3. `Dropout(0.1)`: Uma técnica de regularização que desliga 10% dos neurônios aleatoriamente durante o treino para evitar que o modelo 'decore' os dados (overfitting).

4. Camada Intermediária: Adiciona outra camada com 8 neurônios para extrair mais características.

5. Camada de Saída `(Dense(units=1))`:
- Tem apenas 1 neurônio porque é um problema de classificação binária (ou é spam, ou não é).
- `activation='sigmoid'`: Transforma a saída em uma probabilidade entre 0 e 1.

```python
modelo = Sequential()
modelo.add(Dense(units=10, activation="relu", input_dim=X_train.shape[1]))
modelo.add(Dropout(0.1))
modelo.add(Dense(units=8, activation="relu"))
modelo.add(Dropout(0.1))
modelo.add(Dense(units=1, activation="sigmoid"))
```

```python
modelo.compile(loss="mean_squared_error", optimizer="adam", metrics=["accuracy"])
modelo.summary()
```

```python
modelo.fit(X_train, y_train, epochs=20, batch_size=10, validation_data=(X_test, y_test))
```

```python
loss, accuracy = modelo.evaluate(X_test, y_test)
print("Loss:", loss)
print("Acurácia:", accuracy)
```

```python
nova_previsao = modelo.predict(X_test)
print(nova_previsao)
```

```python
prev2 = (nova_previsao > 0.5)
print(prev2)
```

```python
cm = confusion_matrix(y_test, prev2)
print(cm)
```
