# Vader: Análise de sentimento com regras

```python
import nltk
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer
mas = SentimentIntensityAnalyzer() # 'mas' = minha análise de sentimento
```

```python
a = 'I HATE this movie!!!'
x = mas.polarity_scores(a)
print(x)
```

```python
a = 'I HATE this movie'
x = mas.polarity_scores(a)
print(x)
```

```python
a = 'I hate this movie'
x = mas.polarity_scores(a)
print(x)
```

```python
a = 'I hate this movie! But I loved the girl.'
x = mas.polarity_scores(a)
print(x)
```

```python
a = 'This plot is very old'
x = mas.polarity_scores(a)
print(x)
```

```python
a = ':)'
x = mas.polarity_scores(a)
print(x)
```

```python
a = ':('
x = mas.polarity_scores(a)
print(x)
```

```python
a = 'I hated this movie, but I loved the girl.'
x = mas.polarity_scores(a)
print(x)
```

```python
from textblob import TextBlob
teste = TextBlob('This movie was awesome!')
print(teste.sentiment)
```

```python
from textblob import TextBlob
teste = TextBlob('I love NY!')
print(teste.sentiment)
```

Principais insights:

- **Diferença de intensidade:** o TextBlob consegue distinguir que 'awesome' é mais forte que 'love' em termos de pontuação bruta.
- **Opiniçao vs. fato:** ele ajuda a filtrar o que é uma descrição neutra/fática do que é um julgamento de valor do autor.
