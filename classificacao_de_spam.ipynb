# Classificação de Spam

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.metrics import confusion_matrix, accuracy_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
spam = pd.read_csv("spam.csv")
spam.head()
```

```python
spam.shape
```

```python
spam['Category'].value_counts()
```

```python
distribuicao_percentual = spam['Category'].value_counts(normalize=True)
print(distribuicao_percentual)
```

```python
prev = spam['Message']
classe = spam['Category']
```

```python
vetorizador = TfidfVectorizer()
previsores = vetorizador.fit_transform(prev)
previsores.shape
```

```python
print(vetorizador.get_feature_names_out()[0:100])
```

```python
# `test_size` = percentual de dados que serão usados para o teste,
# o restante será usado para treinamento.

X_treinamento, X_teste, y_treinamento, y_teste = train_test_split(previsores, classe, test_size=0.3)
```

```python
X_teste.shape
```

```python
# `n_estimators` define o número de árvores de decisão da floresta.
# Mais árvores melhoram melhoram a precisão do modelo,
# porém o custo computacional será maior.

floresta = RandomForestClassifier(n_estimators=500)  # o valor padrão do n_estimators é 100
floresta.fit(X_treinamento, y_treinamento)
```

```python
previsoes = floresta.predict(X_teste)
print(previsoes)
```

```python
# y_teste = o que de fato aconteceu
# previsoes = previsao feita pelo modelo

print(confusion_matrix(y_teste, previsoes))
```

**[VP] [FN]**
<br>
**[FP] [VN]**

* **[VP], [VN]** = acertos
* **[FP]** = não eram Spam e o modelo atribuiu como Spam
* **[VN]** = eram spam e o modelo não atribuiu como Spam


```python
print(accuracy_score(y_teste, previsoes))
```

```python
print(metrics.classification_report(y_teste, previsoes))
```
